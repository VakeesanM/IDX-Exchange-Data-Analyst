```python
import pandas as pd
import os
```

```python
path = os.cwd()
```

```python
df = pd.read_csv(path)
```

```python
cols = set(df.columns)
```

```python
df['PriceRation'] = df['ClosePrice']/df['OriginalListPrice']
```

```python
df['PricePerSqft'] = df['ClosePrice']/df['LivingArea']
```

```python
df['Year'] = pd.to_datetime(df['CloseDate']).dt.to_period("Y").head()
```

```python
df['Month'] = pd.to_datetime(df['CloseDate']).dt.month.head()
```

```python
df['YrMo'] = pd.to_datetime(df['CloseDate']).dt.to_period("M").head()
```

```python
df['ListingToContractDay'] = pd.to_datetime(df['CloseDate']) - pd.to_datetime(df['ListingContractDate'])
```

```python
df['ContracttoCloseDays']= pd.to_datetime(df['CloseDate']) - pd.to_datetime(df['PurchaseContractDate'])
```
